```python
%matplotlib inline

import matplotlib.pyplot as plt
import pandas as pd

FS = 250          # Hz. Tanpa nilai ini, sumbu waktu tidak punya makna.
DURASI_PLOT = 5   # detik
```

### Langkah 1 — lihat dulu isi datanya

Kebiasaan yang menyelamatkan banyak waktu: sebelum menghitung apa pun, periksa nama kolom, jumlah baris, dan rentang nilainya. Banyak bug analisis sinyal sebenarnya cuma salah nama kolom atau salah satuan.

```python
frame = pd.read_csv("data/ecg_sample.csv")

print("Kolom yang tersedia :", list(frame.columns))
print("Jumlah baris        :", len(frame))
frame.describe()
```

### Langkah 2 — satu rumus yang perlu dihafal hari ini

$$\text{indeks} = \text{waktu} \times f_s$$

10 detik data @250 Hz = 2.500 sampel. 5 detik pertama = 1.250 sampel pertama. Kalau kamu salah `fs`, seluruh sumbu waktu ikut salah — dan grafiknya tetap terlihat wajar. Itu jenis kesalahan paling berbahaya: tidak ada pesan error.

```python
waktu = frame["time_s"].to_numpy()
ekg = frame["ecg_mv"].to_numpy()

jumlah_sampel = DURASI_PLOT * FS
waktu_window = waktu[:jumlah_sampel]
ekg_window = ekg[:jumlah_sampel]

print("Total sampel   :", ekg.size)
print("Sampel window  :", ekg_window.size)
print("Rentang waktu  : %.2f s .. %.2f s" % (waktu_window[0], waktu_window[-1]))
```

### Langkah 3 — plot, lengkap dengan label

Grafik sinyal medis tanpa label sumbu dan tanpa `fs` di judul praktis tidak bisa dinilai orang lain. Ini bukan soal estetika, tapi soal apakah hasilmu bisa dipertanggungjawabkan.

```python
plt.figure(figsize=(11, 4))
plt.plot(waktu_window, ekg_window, linewidth=1.0)
plt.title(f"EKG sintetis, {DURASI_PLOT} detik pertama (fs = {FS} Hz)")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo (mV)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
